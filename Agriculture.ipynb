```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv("agriculture.csv")
```

```python
print("Shape:", data.shape)

print("\nColumns:")
print(data.columns.tolist())

print("\nFirst 5 rows:")
display(data.head())

print("\nData types:")
print(data.dtypes)

print("\nMissing values:")
print(data.isnull().sum())

print("\nDuplicate rows:", data.duplicated().sum())
```

```python
categorical_cols = [
    "State",
    "District",
    "Crop",
    "Season",
    "Irrigation_Method"
]

for col in categorical_cols:
    print(f"\n===== {col} =====")
    print(data[col].value_counts(dropna=False))
```

```python
numeric_cols = [
    "Farm_Area_Hectares",
    "Rainfall_mm",
    "Avg_Temperature_C",
    "Humidity_pct",
    "Sunlight_Hours_Day",
    "Soil_pH",
    "Soil_Moisture_pct",
    "Nitrogen_kg_ha",
    "Phosphorus_kg_ha",
    "Potassium_kg_ha",
    "Fertilizer_kg_ha",
    "Pesticide_Litre_ha",
    "Seed_Quality_Score",
    "Yield_Tonnes_Ha",
    "Production_Tonnes",
    "Market_Price_INR_Tonne",
    "Total_Cost_INR",
    "Revenue_INR",
    "Profit_INR",
    "Water_Used_m3",
    "Water_Efficiency_t_per_1000m3",
    "Disease_Pest_Risk_pct"
]

display(data[numeric_cols].describe().T)
```

```python
print("Unique Seasons:", data["Season"].unique())
print("Unique Crops:", data["Crop"].unique())
print("Unique States:", data["State"].nunique())
print("Unique Districts:", data["District"].nunique())
print("Unique Irrigation Methods:", data["Irrigation_Method"].unique())

print("\nNegative values:")
print((data[numeric_cols] < 0).sum())
```

```python
print("Revenue = Cost + Profit check:")
print(
    np.isclose(
        data["Total_Cost_INR"] + data["Profit_INR"],
        data["Revenue_INR"]
    ).sum(),
    "out of",
    len(data),
    "rows match"
)

print("\nProduction calculation check:")
print(
    np.isclose(
        data["Farm_Area_Hectares"] * data["Yield_Tonnes_Ha"],
        data["Production_Tonnes"],
        rtol=0.02
    ).sum(),
    "out of",
    len(data),
    "rows approximately match"
)
```

```python
# Missing value summary
missing = data.isnull().sum()
missing = missing[missing > 0].sort_values(ascending=False)

print("Missing values:")
print(missing)

# Rows with missing values
print("\nRows containing missing values:")
display(data[data.isnull().any(axis=1)])

# Yield outliers
print("\nHighest Yield values:")
display(
    data[["Farm_ID", "Crop", "Season", "Farm_Area_Hectares",
          "Yield_Tonnes_Ha", "Production_Tonnes"]]
    .sort_values("Yield_Tonnes_Ha", ascending=False)
    .head(15)
)

# Water efficiency outliers
print("\nHighest Water Efficiency values:")
display(
    data[["Farm_ID", "Crop", "Season", "Yield_Tonnes_Ha",
          "Water_Used_m3", "Water_Efficiency_t_per_1000m3"]]
    .sort_values("Water_Efficiency_t_per_1000m3", ascending=False)
    .head(10)
)
```

```python
# Calculate yield from production and farm area
calculated_yield = data["Production_Tonnes"] / data["Farm_Area_Hectares"]

comparison = pd.DataFrame({
    "Original_Yield": data["Yield_Tonnes_Ha"],
    "Calculated_Yield": calculated_yield
})

comparison["Difference"] = (
    comparison["Original_Yield"] - comparison["Calculated_Yield"]
).abs()

print("Yield comparison:")
display(comparison.describe())

print("\nRows with large yield differences:")
display(
    data.loc[
        comparison["Difference"] > 0.10,
        ["Farm_ID", "Farm_Area_Hectares", "Yield_Tonnes_Ha", "Production_Tonnes"]
    ].head(20)
)
```

```python
yield_by_crop = data.groupby("Crop").agg(
    Count=("Yield_Tonnes_Ha", "count"),
    Mean=("Yield_Tonnes_Ha", "mean"),
    Median=("Yield_Tonnes_Ha", "median"),
    Maximum=("Yield_Tonnes_Ha", "max")
).round(2)

display(yield_by_crop.sort_values("Median", ascending=False))
```

```python
clean_data = data.copy()

# Fill Rainfall using median within each season
clean_data["Rainfall_mm"] = clean_data["Rainfall_mm"].fillna(
    clean_data.groupby("Season")["Rainfall_mm"].transform("median")
)

# Fill Soil Moisture using median within each season
clean_data["Soil_Moisture_pct"] = clean_data["Soil_Moisture_pct"].fillna(
    clean_data.groupby("Season")["Soil_Moisture_pct"].transform("median")
)

# Fill missing Yield using Production / Farm Area
clean_data["Yield_Tonnes_Ha"] = clean_data["Yield_Tonnes_Ha"].fillna(
    clean_data["Production_Tonnes"] / clean_data["Farm_Area_Hectares"]
)

print("Missing values after cleaning:")
print(clean_data.isnull().sum()[clean_data.isnull().sum() > 0])

print("\nCleaned dataset shape:", clean_data.shape)
```

```python
print("Original rows:", len(data))
print("Cleaned rows:", len(clean_data))
print("Original duplicates:", data.duplicated().sum())
print("Cleaned duplicates:", clean_data.duplicated().sum())

print("\nMissing values remaining:")
print(clean_data.isnull().sum().sum())
```

```python
from scipy.stats import f_oneway

kharif = clean_data.loc[clean_data["Season"] == "Kharif", "Yield_Tonnes_Ha"]
rabi = clean_data.loc[clean_data["Season"] == "Rabi", "Yield_Tonnes_Ha"]
zaid = clean_data.loc[clean_data["Season"] == "Zaid", "Yield_Tonnes_Ha"]

f_stat, p_value = f_oneway(kharif, rabi, zaid)

print("ANOVA F-statistic:", round(f_stat, 4))
print("p-value:", p_value)

if p_value < 0.05:
    print("Result: Yield differs significantly across seasons.")
else:
    print("Result: No statistically significant difference in yield across seasons.")
```

```python
kharif_profit = clean_data.loc[clean_data["Season"] == "Kharif", "Profit_INR"]
rabi_profit = clean_data.loc[clean_data["Season"] == "Rabi", "Profit_INR"]
zaid_profit = clean_data.loc[clean_data["Season"] == "Zaid", "Profit_INR"]

f_stat_profit, p_value_profit = f_oneway(
    kharif_profit,
    rabi_profit,
    zaid_profit
)

print("ANOVA F-statistic:", round(f_stat_profit, 4))
print("p-value:", p_value_profit)

if p_value_profit < 0.05:
    print("Result: Profit differs significantly across seasons.")
else:
    print("Result: No statistically significant difference in profit across seasons.")
```
